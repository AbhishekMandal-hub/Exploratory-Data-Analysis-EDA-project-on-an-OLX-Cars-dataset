```python
# finding unique values in car features

i = 1
k = set(df['Car Features'].str.split(",").str[0].unique())
while len(df['Car Features'].str.split(",").str[i].unique()) !=1:
    k = k.union(df['Car Features'].str.split(",").str[i].unique())
    i = i+1
```

```python
k
```

```python
k = [i for i in k if type(i)==str]
k = [i.strip() for i in k ]

def encoder(x):
    t = []
    for i in k:
        if i in x:
            t.append(1)
        else:
            t.append(0)
    return t
```

```python
 df[k] = list(map(encoder,df['Car Features']))
```
